```python
#import sklearn #kullanacağımız kütüphaneyi bu şekilde yükleyebiliriz
#from sklearn ->bu da bir kütüphane içinden bir parça çağırmak
#from sklearn.linear_model import LinearRegression-> kütüphane içindeki bir modelden bir parça çağırma

import pandas as pd #pandas yerine pd diyerek kullanabiliriz kodlarken
from sklearn.linear_model import LinearRegression
#kaggle girip aramaya regression yaz database sekmesinden uygun veri setini indir ve colabda dosyalar kısmına ekle
#.csv uzantılı bir dosya mesela (student.csv olarak kaydettim)

df=pd.read_csv("Student.csv") #cvs dosyasını oku
df.head(3)# 3 satır getir

# Kategorik veriyi (Yes/No) sayısal verilere (1/0) dönüştürüyoruz
df['Extracurricular Activities'] = df['Extracurricular Activities'].map({'Yes': 1, 'No': 0})

y=df[["Performance Index"]] #y bizim tahmin etmek istediğimiz şey
x=df[["Hours Studied","Previous Scores","Extracurricular Activities","Sleep Hours","Sample Question Papers Practiced"]]#bunlar da tahminimizi etkileyen bağımsız değişkenler

df.info()#data hakkında bilgi alabiliriz -> kaç satırdan oluşuyo,stunların veri tipi vs

L=LinearRegression()#yukarda import etmiştim

#Model Nesnesi Oluşturur: L adında boş bir lineer regresyon modeli
#tanımlar. Bu aşamada henüz verileri incelememiştir;
#sadece modeli kurmak için gerekli altyapıyı hazırlar

model=L.fit(x,y)# Küçük l yerine büyük L kullandık.
#Model bu katsayıları bulmak için "En Küçük Kareler" (Least Squares) adı verilen matematiksel bir yöntem kullanır.
#Kısacası fit() komutu, verilere en uygun çizgiyi/denklemi çizebilmek için gereken o en doğru $a$ ve $b$ katsayılarını hesaplayan matematiksel bir optimizasyon yapar

# 'Yes' yerine sayısal karşılığı olan 1 değerini verdik
model.predict([[5, 88, 1, 6, 4]]) #tahmin et dedim, neyi tahmin edeceğini () içine yaz

```

```python
model.score(x,y)#ne oranda doğru tahmin etmişim
```
